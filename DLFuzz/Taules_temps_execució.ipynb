```python
from pathlib import Path
import re
import pandas as pd

FOLDER = Path("/content/TFG/DLFuzz/Resultats_imagenet")

rows = []

for txt in FOLDER.glob('*.txt'):
    with open(txt, 'r', encoding='utf-8') as fh:
        for line in reversed(fh.readlines()):
            if line.strip().startswith('total_time'):
                total_time = float(line.split('=')[1])
                break

    m = re.search(r'(\d{2})(?=\.txt$)', txt.name)
    if m is None:
        raise ValueError(f"No s’han trobat dos dígits al final del nom: {txt.name}")

    digits = m.group(1)
    model     = int(digits[0])
    strategy  = int(digits[1]) + 1

    rows.append({'Model': model, 'Strategy': strategy, 'Total_time': total_time})


df = pd.DataFrame(rows)
table = (
    df
    .pivot(index='Model', columns='Strategy', values='Total_time')
    .sort_index()
    .sort_index(axis=1)
)


display(table)
```
