```python
import os
import shutil

def copy_all_files(src_root, dest_root):
    for root, dirs, files in os.walk(src_root):
        for file in files:
            src_file = os.path.join(root, file)
            dest_file = os.path.join(dest_root, file)

            if os.path.exists(dest_file):
                base, ext = os.path.splitext(file)
                count = 1
                while os.path.exists(dest_file):
                    dest_file = os.path.join(dest_root, f"{base}_{count}{ext}")
                    count += 1

            shutil.copy2(src_file, dest_file)

folderA = '/content/Other_signs/Other_countries'
folderB = '/content/Other_signs/Other_type'

outputA = 'altres_senyals/altres_paisos'
outputB = 'altres_senyals/altres_tipus'

os.makedirs(outputA, exist_ok=True)
os.makedirs(outputB, exist_ok=True)

copy_all_files(folderA, outputA)
copy_all_files(folderB, outputB)
```

```python
def show_output(log_path):
    with open(log_path, 'r') as f:
        for line in f:
            line = line.strip()
            if line.startswith("SHOW_IMAGE:"):
                path = re.search(r'SHOW_IMAGE:\s*(.*)', line).group(1)
                display(Image(filename=path))
            else:
                print(line)
```

```python
from IPython.display import Image, display, HTML
import re

# --- Helper: Parse one log file into a list of entries ---
def parse_log_file(filepath):
    entries = []
    current = None

    with open(filepath, 'r') as f:
        for line in f:
            line = line.strip()
            if line.startswith('Image path:'):
                if current:
                    entries.append(current)
                current = {
                    'image_path': line.split(': ', 1)[1],
                    'output': [line]
                }
            elif line.startswith('SHOW_IMAGE:'):
                if current:
                    current['image_display_path'] = line.split(': ', 1)[1]
                    current['output'].append(line)
            else:
                if current:
                    current['output'].append(line)
                else:
                    # Ignore lines before the first image path
                    continue

    if current:
        entries.append(current)

    return entries

def remove_ansi_escape_codes(text):
    return re.sub(r'\x1b\[[0-9;]*m', '', text)

def remove_unwanted_lines(text):
    cleaned_text = remove_ansi_escape_codes(text)
    lines = cleaned_text.split('\n')
    filtered_lines = [line for line in lines if not line.startswith("Models predictions:")]
    return '\n'.join(filtered_lines)

def get_filename(path):
    return path.split('/')[-1]
```

```python
trained_entries = parse_log_file('/content/output_log1.txt')
retrained_entries = parse_log_file('/content/output_log2.txt')

trained_dict = {get_filename(e['image_path']): e for e in trained_entries}
retrained_dict = {get_filename(e['image_path']): e for e in retrained_entries}

for filename in sorted(set(trained_dict) & set(retrained_dict)):
    trained = trained_dict[filename]
    retrained = retrained_dict[filename]

    display(HTML(f"<h4>Image: {filename}</h4>"))

    display(Image(filename=trained['image_display_path'], width=300))

    trained_output = remove_unwanted_lines('\n'.join(trained['output'][2:]))
    retrained_output = remove_unwanted_lines('\n'.join(retrained['output'][2:]))

    trained_output = re.sub(r"(Top 5 Predictions:)", r'<span style="color:blue;">\1</span>', trained_output)
    retrained_output = re.sub(r"(Top 5 Predictions:)", r'<span style="color:blue;">\1</span>', retrained_output)

    html = f"""
    <table style="width:100%; table-layout:fixed; border-collapse:collapse; margin-bottom:30px;">
        <tr>
            <th style="width:50%; text-align:left; padding:5px; border-bottom:1px solid #ccc;">Trained Model</th>
            <th style="width:50%; text-align:left; padding:5px; border-bottom:1px solid #ccc;">Retrained Model</th>
        </tr>
        <tr>
            <td style="vertical-align:top; padding:5px;"><pre>{trained_output}</pre></td>
            <td style="vertical-align:top; padding:5px;"><pre>{retrained_output}</pre></td>
        </tr>
    </table>
    """
    display(HTML(html))
```
