```python
import pandas as pd
import os
import glob

def find_txt_files(ending_number, initial_path=""):
    """
    Find all .txt files in the specified directory (or current directory if not given) that end with the specified number (0, 1, or 2).
    """
    search_path = os.path.join(initial_path, "*{0}.txt".format(ending_number))

    return glob.glob(search_path)

def extract_image_mappings(txt_files):
    """
    Extracts image mappings from multiple txt files.
    Returns a dictionary {original_path: [new_path1, new_path2, ...]}.
    """
    mappings = {}  # Store original -> list of new paths

    for txt_file in txt_files:
        with open(txt_file, 'r') as file:
            lines = file.readlines()

        dataset_name = txt_file.replace("/content/TFG/DeepXplore/GTSRB/Models_123/Resultats_DeepXplore_GTSRB/output_", "generated_inputs_").replace(".txt", "")
        #print(dataset_name)
        i = 0
        while i < len(lines) - 2:
            #print(lines[i])
            if lines[i].startswith("Image path:") and lines[i + 1].startswith("Models predictions before:"):
                original_path = lines[i].split("Image path:")[1].strip()
                #print(original_path)
                for j in range(i + 2, len(lines)):
                    if lines[j].strip().startswith("./generated_inputs/"):
                        #print(lines[j])
                        new_path = lines[j].strip().replace("./generated_inputs", f"{dataset_name}")
                        #print(new_path)
                        if original_path in mappings:
                            mappings[original_path].append(new_path)
                        else:
                            mappings[original_path] = [new_path]
                        break
            i += 1

    return mappings

def update_dataset(csv_file, mappings, output_file, new_test_file):
    """
    Loads dataset, replaces original paths with new paths, and removes old rows.
    Returns two datasets:
    - One with the new rows.
    - One with the original dataset excluding the removed rows.
    """
    df = pd.read_csv(csv_file)

    new_rows = []
    original_paths_to_remove = set()

    for original_path, new_paths in mappings.items():
        if original_path in df["Path"].values:
            original_paths_to_remove.add(original_path)
            for new_path in new_paths:
                row = df[df["Path"] == original_path].copy()
                row["Path"] = new_path
                new_rows.append(row)

    new_rows_df = pd.concat(new_rows, ignore_index=True)
    new_rows_df.to_csv(output_file, index=False)

    test_df = df[~df["Path"].isin(original_paths_to_remove)]

    test_df.to_csv(new_test_file, index=False)

    return output_file, new_test_file

folder = "/content/TFG/DeepXplore/GTSRB/Models_123/Resultats_DeepXplore_GTSRB/"

txt_files = glob.glob(os.path.join(folder, "*.txt"))

image_mappings = extract_image_mappings(txt_files)

csv_file = '/content/gtsrb-german-traffic-sign/Test.csv'
output_file = "/content/Model_tot_retrain_dataset.csv"
new_test_file = "/content/Model_tot_test_df.csv"

retrain_df, test_df = update_dataset(csv_file, image_mappings, output_file, new_test_file)
print(retrain_df)
```
